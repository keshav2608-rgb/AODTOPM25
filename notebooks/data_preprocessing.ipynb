```python
import netCDF4 as nc
import numpy as np
import pandas as pd
import os
from datetime import datetime, timedelta
import pyarrow as pa
import pyarrow.parquet as pq

DATA_DIR = r"C:\Users\kesha\OneDrive - dtu.ac.in\Desktop\aodtopm25\data\raw\merra"
OUTPUT_FILE = r"C:\Users\kesha\OneDrive - dtu.ac.in\Desktop\aodtopm25\data\interim\merra_merged.parquet"

VARIABLES = ["PBLH", "TLML", "QLML", "ULML", "VLML", "SPEED", "PRECTOT"]

files = sorted([f for f in os.listdir(DATA_DIR) if f.endswith(".nc4")])
print(f"Found {len(files)} files\n")

parquet_writer = None

for i, fname in enumerate(files):
    try:
        ds = nc.Dataset(os.path.join(DATA_DIR, fname))

        lons = ds.variables["lon"][:]
        lats = ds.variables["lat"][:]
        times = ds.variables["time"]

        # Time parsing
        time_units = times.units
        base_time_str = time_units.split("since ")[-1]
        base_time = datetime.strptime(base_time_str.strip(), "%Y-%m-%d %H:%M:%S")
        timestamps = [base_time + timedelta(minutes=int(t)) for t in times[:]]

        lon2d, lat2d = np.meshgrid(lons, lats)
        lon_flat = lon2d.flatten()
        lat_flat = lat2d.flatten()

        for t_idx, timestamp in enumerate(timestamps):
            data_dict = {
                "datetime": np.full_like(lon_flat, timestamp, dtype=object),
                "latitude": lat_flat,
                "longitude": lon_flat,
            }

            for var in VARIABLES:
                if var in ds.variables:
                    data = ds.variables[var][t_idx, :, :].filled(np.nan).flatten()
                    data_dict[var] = data

            df = pd.DataFrame(data_dict)

            table = pa.Table.from_pandas(df)

            # Initialize writer only once
            if parquet_writer is None:
                parquet_writer = pq.ParquetWriter(OUTPUT_FILE, table.schema)

            parquet_writer.write_table(table)

        ds.close()
        print(f"[{i+1}/{len(files)}] {fname} ✅")

    except Exception as e:
        print(f"[{i+1}/{len(files)}] {fname} ❌ {e}")

# Close writer
if parquet_writer:
    parquet_writer.close()

print(f"\n✅ Single Parquet saved at: {OUTPUT_FILE}")
```

```python
import pandas as pd
import numpy as np

# =========================
# 1. LOAD DATA
# =========================
# Change this to your actual file
merra_df = pd.read_parquet(r"C:\Users\kesha\OneDrive - dtu.ac.in\Desktop\aodtopm25\data\interim\india_clipped\merra_india.parquet")
# or:
# merra_df = pd.read_csv("merra_data.csv")

# =========================
# 2. DATETIME FEATURES
# =========================
merra_df['datetime'] = pd.to_datetime(merra_df['datetime'])
merra_df['hour'] = merra_df['datetime'].dt.hour

# =========================
# 3. WIND FEATURES
# =========================
# Wind direction from U and V components
merra_df['wind_dir'] = np.degrees(
    np.arctan2(merra_df['VLML'], merra_df['ULML'])
) % 360

# Keep raw components
merra_df['wind_u'] = merra_df['ULML']
merra_df['wind_v'] = merra_df['VLML']

# =========================
# 4. ATMOSPHERIC FEATURES
# =========================
# Stability / inversion proxy
merra_df['inversion_proxy'] = merra_df['TLML'] / (merra_df['PBLH'] + 1)

# Ventilation coefficient
merra_df['vent_coeff'] = merra_df['PBLH'] * merra_df['SPEED']

# Moisture (specific humidity → g/kg)
merra_df['moisture'] = merra_df['QLML'] * 1000

# =========================
# 5. SAVE PROCESSED DATA
# =========================

# Best option (recommended)
merra_df.to_parquet(r"C:\Users\kesha\OneDrive - dtu.ac.in\Desktop\aodtopm25\data\interim\india_clipped\merra_features.parquet", index=False)

# Optional: also save CSV (slower, larger)
# merra_df.to_csv("merra_features.csv", index=False)

print("✅ Feature engineering complete and file saved!")
```

```python
import pyarrow.parquet as pq

file_path = r"C:\Users\kesha\OneDrive - dtu.ac.in\Desktop\aodtopm25\data\interim\merra_daily\merra_daily_grid.parquet"
parquet_file = pq.ParquetFile(file_path)


print("Number of row groups:", parquet_file.num_row_groups)
print("Schema:")
print(parquet_file.schema)
```

```python
min_lat, max_lat = 999, -999
min_lon, max_lon = 999, -999

for i in range(parquet_file.num_row_groups):
    df = parquet_file.read_row_group(i, columns=["lat_round", "lon_round"]).to_pandas()
    
    min_lat = min(min_lat, df["lat_round"].min())
    max_lat = max(max_lat, df["lat_round"].max())
    min_lon = min(min_lon, df["lon_round"].min())
    max_lon = max(max_lon, df["lon_round"].max())

print("Latitude Min:", min_lat)
print("Latitude Max:", max_lat)
print("Longitude Min:", min_lon)
print("Longitude Max:", max_lon)
```
